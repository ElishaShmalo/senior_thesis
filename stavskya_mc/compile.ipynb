```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl

def set_default_plot_style():
    plt.style.use("default")  # start clean

    plt.rcParams["axes.grid"] = True
    plt.rcParams["grid.alpha"] = 0.2
    plt.rcParams["grid.linestyle"] = "-"
    plt.rcParams["grid.color"] = "gray"

    plt.rcParams["axes.spines.top"] = False
    plt.rcParams["axes.spines.right"] = False

    plt.rcParams["axes.linewidth"] = 1.2

    plt.rcParams["legend.frameon"] = False

    plt.rcParams["font.size"] = 16
    plt.rcParams["axes.labelsize"] = 24
    plt.rcParams["legend.fontsize"] = 18

    mpl.rcParams['text.usetex'] = True
    mpl.rcParams['font.family'] = 'serif'
    mpl.rcParams['font.serif'] = ['Computer Modern Roman']

# Call once at the top of your script
set_default_plot_style()
```

## The complied critical values

```python
delta_randomness_results = { # delta: {p_c, delta, nu_par, z, beta, nu_perp}
    0.1: {"epsilon_c": 0.28775, "delta": 0.15, "nu_par":1.67, "z": 1.32, "beta": 0.2505, "nu_perp": 1.265},
    0.2: {"epsilon_c": 0.26779, "delta": 0.13, "nu_par":1.875, "z": 1.35, "beta": 0.244, "nu_perp": 1.39},
    0.3: {"epsilon_c": 0.252325, "delta": 0.101, "nu_par":2.3, "z": 1.4, "beta": 0.23, "nu_perp": 1.643},
    0.35: {"epsilon_c": 0.25233, "delta": 0.101, "nu_par":2.1, "z": None, "beta": 0.21, "nu_perp": None},
}
```

```python
def draw_line_no_scale(ax, xs, ys):
    xlim = ax.get_xlim()
    ylim = ax.get_ylim()

    ax.plot(xs, ys,
            c="k", linestyle="dashed")

    ax.set_xlim(xlim)
    ax.set_ylim(ylim)
```

## Plotting the critical values vs Delta

```python
# p_c
fig, ax = plt.subplots()
ax.plot(
    delta_randomness_results.keys(), [delta_randomness_results[delta_val]["epsilon_c"] for delta_val in delta_randomness_results], 
    c="r", label=r"$\varepsilon_c$", marker="o")
ax.plot(
    [min(delta_randomness_results[delta_val]["epsilon_c"], delta_val) for delta_val in delta_randomness_results], [delta_randomness_results[delta_val]["epsilon_c"] for delta_val in delta_randomness_results], 
    c=(1, 0, 0, 0.5), marker="o")
draw_line_no_scale(ax, [0, 1], [0, 1])
ax.legend()
plt.xlabel(r"$\Delta$")
plt.ylabel(r"$\varepsilon_c$")
plt.show()

# delta
plt.plot(
    [min(delta_randomness_results[delta_val]["epsilon_c"], delta_val) for delta_val in delta_randomness_results], [delta_randomness_results[delta_val]["delta"] for delta_val in delta_randomness_results], 
    c=(0, 1, 0, 0.5), marker="o")
plt.plot(
    delta_randomness_results.keys(), [delta_randomness_results[delta_val]["delta"] for delta_val in delta_randomness_results], 
    c="g", label=r"$\delta$", marker="o")
plt.legend()
plt.xlabel(r"$\Delta$")
plt.ylabel(r"$\delta$")
plt.show()

# nu_par
plt.plot(
    [min(delta_randomness_results[delta_val]["epsilon_c"], delta_val) for delta_val in delta_randomness_results], [delta_randomness_results[delta_val]["nu_par"] for delta_val in delta_randomness_results], 
    c=(0, 0, 1, 0.5), marker="o")
plt.plot(
    delta_randomness_results.keys(), [delta_randomness_results[delta_val]["nu_par"] for delta_val in delta_randomness_results], 
    c="b", label=r"$\nu_{\|}$", marker="o")
plt.legend()
plt.xlabel(r"$\Delta$")
plt.ylabel(r"$\nu_{\|}$")
plt.show()

# z
plt.plot(
    [min(delta_randomness_results[delta_val]["epsilon_c"], delta_val) for delta_val in delta_randomness_results], [delta_randomness_results[delta_val]["z"] for delta_val in delta_randomness_results], 
    c=(0, 1, 1, 0.5), marker="o")
plt.plot(
    delta_randomness_results.keys(), [delta_randomness_results[delta_val]["z"] for delta_val in delta_randomness_results], 
    c="c", label=r"$z$", marker="o")
plt.legend()
plt.xlabel(r"$\Delta$")
plt.ylabel(r"$z$")
plt.show()

# beta
plt.plot(
    [min(delta_randomness_results[delta_val]["epsilon_c"], delta_val) for delta_val in delta_randomness_results], [delta_randomness_results[delta_val]["beta"] for delta_val in delta_randomness_results], 
    c=(1, 0, 1, 0.5), marker="o")
plt.plot(
    delta_randomness_results.keys(), [delta_randomness_results[delta_val]["beta"] for delta_val in delta_randomness_results], 
    c="m", label=r"$\beta$", marker="o")
plt.legend()
plt.xlabel(r"$\Delta$")
plt.ylabel(r"$\beta$")
plt.show()

# nu_perp
plt.plot(
    [min(delta_randomness_results[delta_val]["epsilon_c"], delta_val) for delta_val in delta_randomness_results], [delta_randomness_results[delta_val]["nu_perp"] for delta_val in delta_randomness_results], 
    c=(1, 0.75, 0, 0.5), marker="o")
plt.plot(
    delta_randomness_results.keys(), [delta_randomness_results[delta_val]["nu_perp"] for delta_val in delta_randomness_results], 
    c=(1, 0.75, 0, 1), label=r"$\nu_{\perp}$", marker="o")
plt.legend()
plt.xlabel(r"$\Delta$")
plt.ylabel(r"$\nu_{\perp}$")
plt.show()
```
